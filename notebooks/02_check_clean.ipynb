```python
import pandas as pd

df = pd.read_parquet("../data/processed/retail_clean.parquet")
print(df.shape)
print(df.dtypes)
df.head()
```

```python
orders_per_customer = df.groupby("CustomerID")["InvoiceNo"].nunique()
print("Клиентов всего:            ", len(orders_per_customer))
print("Клиентов с >= 5 заказов:   ", (orders_per_customer >= 5).sum())
print("Клиентов с >= 10 заказов:  ", (orders_per_customer >= 10).sum())
print("Клиентов с >= 20 заказов:  ", (orders_per_customer >= 20).sum())
```

```python
df_sorted = df.sort_values(["CustomerID", "InvoiceDate"])
is_sorted = df.equals(df_sorted)
print("Отсортировано правильно:", is_sorted)
```
