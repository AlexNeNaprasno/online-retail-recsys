```python
import pandas as pd
import json

train = pd.read_parquet("../data/processed/splits/train.parquet")
val   = pd.read_parquet("../data/processed/splits/val.parquet")
test  = pd.read_parquet("../data/processed/splits/test.parquet")

with open("../data/processed/vocab.json", encoding="utf-8") as f:
    vocab = json.load(f)

inv_vocab = {v: k for k, v in vocab.items()}

print("Train:", train.shape)
print("Val:  ", val.shape)
print("Test: ", test.shape)

# Первый клиент
cid = train.iloc[0]["customer_id"]
print(f"\nКлиент {cid}:")
print("  train len:", len(train[train.customer_id == cid].iloc[0]["sequence"]))
print("  val   len:", len(val[val.customer_id == cid].iloc[0]["sequence"]))
print("  test  len:", len(test[test.customer_id == cid].iloc[0]["sequence"]))

# Первые 10 элементов test в виде товаров
seq = test[test.customer_id == cid].iloc[0]["sequence"]
print("\n  Первые 10 StockCode клиента (test):")
print("  ", [inv_vocab.get(i, "<UNK>") for i in seq[:10]])
```
